```python
import matplotlib.pyplot as plt
import seaborn as sns

# Reshape from wide to long format
melted_data = state_data.melt(
    id_vars="state",
    value_vars=["N", "P", "K", "pH"],
    var_name="Feature",
    value_name="Value",
)

plt.figure(figsize=(12, 6))
sns.lineplot(data=melted_data, x="state", y="Value", hue="Feature", marker="o")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
